# تحليل استكشافي — PhysioNet Sepsis Challenge 2019
تحليل توزيع المتغيرات الحيوية، نسبة حالات الإنتان، والقيم المفقودة.

```python
import sys, os
sys.path.append('../src')
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from preprocessing import load_patient_files, read_patient, FEATURE_COLS, LABEL_COL

files = load_patient_files('../data/raw')
print(f'عدد ملفات المرضى: {len(files)}')
```

```python
sample = pd.concat([read_patient(f) for f in files[:200]], ignore_index=True)
print('نسبة الإنتان (عينة):', sample[LABEL_COL].mean())
sample[FEATURE_COLS].isna().mean().sort_values(ascending=False).plot(kind='bar', figsize=(14,4), title='نسبة القيم المفقودة لكل متغير')
plt.tight_layout(); plt.show()
```

```python
fig, axes = plt.subplots(1, 3, figsize=(15,4))
for ax, col in zip(axes, ['HR', 'MAP', 'Resp']):
    sample.groupby(LABEL_COL)[col].plot(kind='hist', alpha=0.5, ax=ax, legend=True, bins=30, title=col)
plt.tight_layout(); plt.show()
```
